# FIRMS fire detections

Load the same NASA FIRMS feed as `main.py` and inspect the columns. `acq_date` and `frp` are already shown in the map popup.

```python
import os
import requests
import pandas as pd
from io import StringIO
from dotenv import load_dotenv

load_dotenv()
MAP_KEY = os.environ["FIRMS_MAP_KEY"]
url = f"https://firms.modaps.eosdis.nasa.gov/api/area/csv/{MAP_KEY}/MODIS_NRT/-125,24,-66,50/2"

response = requests.get(url)
response.raise_for_status()
df = pd.read_csv(StringIO(response.text))
print(f"{len(df):,} detections")
df.head()
```
